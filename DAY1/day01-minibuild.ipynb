```python
from dotenv import load_dotenv
from openai import OpenAI
```

```python
load_dotenv()
```

```python
client = OpenAI()
```

```python
user_input = input("ASK GPT...")
```

```python
user_input
```

```python
response = client.responses.create(
    model="gpt-6-luna",
    input=user_input,
    instructions="""
        You are a helpful Data Engineering learning assistant.
        Answer questions clearly and concisely.
    """
)
```

```python
print(response.output_text)
```
