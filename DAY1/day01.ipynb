```python
from dotenv import load_dotenv
from openai import OpenAI
```

```python
load_dotenv()
```

```python
client = OpenAI()
```

```python
response = client.responses.create(
    model="gpt-6-luna",
    input="What is a data pipeline?"
)
```

```python
print(response.output_text)
```

### Experiment 1 — No System Instruction

```python
response_1 = client.responses.create(
    model="gpt-6-luna",
    input="What are the top 5 customers by revenue?"
)
```

```python
print(response_1.output_text)
```

### Experiment 2 — Add a System Instruction

```python
response_2 = client.responses.create(
    model="gpt-6-luna",
    instructions="You are a data analyst assistant.",
    input="What are the top 5 customers by revenue?"
)
```

```python
print(response_2.output_text)
```

### Experiment 3 — Explicit Guardrail

```python
response_3 = client.responses.create(
    model="gpt-6-luna",
    instructions="""
    You are a data analyst assistant.
    Never invent data or numerical results.
    If the required data has not been provided,
    explain that the analysis cannot be performed.
    """,
    input="What are the top 5 customers by revenue?"
)
```

```python
print(response_3.output_text)
```
