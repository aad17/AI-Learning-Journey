```python
from dotenv import load_dotenv
from openai import OpenAI
```

```python
load_dotenv()
```

```python
client = OpenAI()
```

```python
user_inputs = {
    "Two records have customer_id 1042.": "DUPLICATE",
    "A record has no customer_id.": "MISSING_VALUE",
    "A record has age 145.": "INVALID_VALUE",
    "A record has customer_id 'ABC'.": "SCHEMA_ERROR",
    "A record has age 45 and no email.": "VALID",
    "Something is wrong with a customer record.": "AMBIGUOUS",
    "Customer address contains an old postcode.": "OTHER",
    "Customer email is abc.example.com.": "INVALID_VALUE"
}
```

```python
for user_input in user_inputs:
    print(user_input)
```

```python
ans = []
```

```python
for user_input in user_inputs:
    response = client.responses.create(
        model="gpt-6-luna",
        input=user_input,
        instructions="""
            TASK:
            Classify the incident into exactly one category.
        
            CATEGORIES:
            - VALID
            - DUPLICATE
            - MISSING_VALUE
            - INVALID_VALUE
            - SCHEMA_ERROR
            - OTHER
            - AMBIGUOUS
        
            RULES:
            - DUPLICATE: a value that should be unique appears more than once.
            - MISSING_VALUE: a required value is absent.
            - INVALID_VALUE: a value violates a known business rule.
            - SCHEMA_ERROR: an unexpected field or datatype is present.
            - OTHER: enough information exists to understand the issue, but it does not match the categories above.
            - AMBIGUOUS: there is not enough information to reliably determine the issue.
            - DO NOT assume a feild is missing if not provided. Only evaluate the feilds that are explicitly mentioned in question.

            BUSINESS CONTEXT:
            - customer_id:
                - required
                - must be unique
                - must be an integer
            - email
                - optional
                - if present, must contain @
            - age
                - required
                - integer
                - valid range: 0-120 inclusive
        
            Return only the category.
        """
    )
    print(f'User Input: {user_input} \nOutput:{response.output_text}')
    ans.append([user_input, response.output_text])
```
