```python
from dotenv import load_dotenv
from openai import OpenAI
```

```python
load_dotenv()
```

```python
client = OpenAI()
```

```python
user_input = input("ASK GPT...")
```

```python
user_input
```

```python
response = client.responses.create(
    model="gpt-6-luna",
    input=user_input,
    instructions="""
        Classify the following data quality incident into exactly one category:

        DUPLICATE
        MISSING_VALUE
        INVALID_VALUE
        SCHEMA_ERROR
        OTHER

        Return only the category name.

        Context:
        - customer_id must be unique.
        - email may be NULL.
        - age is valid between 0 and 150 inclusive.
    """
)
```

```python
response_1 = client.responses.create(
    model="gpt-6-luna",
    input="""
    TASK:
    Classify the incident into exactly one category.

    CATEGORIES:
    - DUPLICATE
    - MISSING_VALUE
    - INVALID_VALUE
    - SCHEMA_ERROR
    - OTHER
    - AMBIGUOUS

    RULES:
    - DUPLICATE: a value that should be unique appears more than once.
    - MISSING_VALUE: a required value is absent.
    - INVALID_VALUE: a value violates a known business rule.
    - SCHEMA_ERROR: an unexpected field or datatype is present.
    - OTHER: enough information exists to understand the issue, but it does not match the categories above.
    - AMBIGUOUS: there is not enough information to reliably determine the issue.

    Return only the category.

    INCIDENT:
    There is a problem with customer data.
    """,
)
```
