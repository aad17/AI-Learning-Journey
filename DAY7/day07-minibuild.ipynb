```python
test_cases = [
    {
        "id": 1,
        "question": "Classify urgent incident",
        "expected": "HIGH",
        "actual": "HIGH",
        "expected_format": "CATEGORY",
    },
    {
        "id": 2,
        "question": "Classify low priority incident",
        "expected": "LOW",
        "actual": "Priority is LOW",
        "expected_format": "CATEGORY",
    },
    {
        "id": 3,
        "question": "Calculate total revenue",
        "expected": 50000,
        "actual": 45000,
        "expected_format": "NUMBER",
    },
    {
        "id": 4,
        "question": "Classify medium priority incident",
        "expected": "MEDIUM",
        "actual": "MEDIUM",
        "expected_format": "CATEGORY",
    },
    {
        "id": 5,
        "question": "Calculate completed order count",
        "expected": 120,
        "actual": 120,
        "expected_format": "NUMBER",
    },
]
```

```python
def evaluate_response(test_case: dict) -> dict:
    """
    Compare actual output against expected output.

    Return:
    {
        "id": int,
        "passed": bool,
        "failure_category": str | None
    }
    """
    # TODO: Implement evaluation logic
    passed = test_case["expected"] == test_case["actual"]
    if test_case["expected_format"] == "CATEGORY":
        return {
            "id": test_case["id"],
            "passed": passed,
            "failure_category": None if passed else "FORMAT_FAILURE"
        }
    elif test_case["expected_format"] == "NUMBER":
        return {
            "id": test_case["id"],
            "passed": passed,
            "failure_category": None if passed else "LOGICAL_FAILURE"
        }
```

```python
def run_evaluation(cases):
    results = []

    for case in cases:
        result = evaluate_response(case)
        results.append(result)

    return results
```

```python
def print_report(results):
    total = len(results)
    passed = sum(r["passed"] for r in results)
    accuracy = (passed / total) * 100 if total else 0

    print(f"Total Tests: {total}")
    print(f"Passed: {passed}")
    print(f"Failed: {total - passed}")
    print(f"Accuracy: {accuracy:.2f}%")

    print("\nDetailed Results:")
    for result in results:
        print(result)
```

```python
results = run_evaluation(test_cases)
print_report(results)
```

```python
v1_results = [
    {"id": 1, "passed": True},
    {"id": 2, "passed": False},
    {"id": 3, "passed": True},
    {"id": 4, "passed": True},
    {"id": 5, "passed": False},
]

v2_results = [
    {"id": 1, "passed": True},
    {"id": 2, "passed": True},
    {"id": 3, "passed": False},
    {"id": 4, "passed": True},
    {"id": 5, "passed": True},
]
```

```python
def compare_versions(v1_results: list, v2_results: list) -> dict:
    """
    Compare evaluation results.

    Return:
    {
        "improvements": [test IDs],
        "regressions": [test IDs],
        "unchanged": [test IDs]
    }
    """
    # Your implementation
    improvements, regressions, unchanged = [], [], []

    ans = {}
    for d in v1_results:
        ans[d["id"]] = d["passed"]

    for d in v2_results:
        if d["id"] in ans:
            if ans[d["id"]] == d["passed"]:
                unchanged.append(d["id"])
            elif ans[d["id"]] == False and d["passed"] == True:
                improvements.append(d["id"])
            elif ans[d["id"]] == True and d["passed"] == False:
                regressions.append(d["id"])

    return {
        "improvements": improvements,
        "regressions": regressions,
        "unchanged": unchanged
    }
```

```python
compare_versions(v1_results, v2_results)
```
